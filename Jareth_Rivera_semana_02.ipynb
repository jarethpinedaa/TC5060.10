*   **Pregunta 1.**

Busca y elimina todos los saltos de línea '\n', así como las etiquetas de 0s y 1s que se encuentran al final de cada comentario. En esta actividad solo trabajaremos con la limpieza de datos, por lo que no requeriremos dichas etiquetas por el momento.

Una vez finalizado, imprime los primeros 10 comentarios del resultado obtenido.


```python
# eliminar \n y etiquetas (0 o 1 al final)
clean_docs = [re.sub(r'\t[01]\n$', '', doc) for doc in docs]

# primeros 10
print(clean_docs[:10])
```

**NOTA: A partir de ahora, deberás trabajar con los comentarios obtenidos de esta primer pregunta.**

*   **Pregunta 2.**  

Busca e imprime todas las palabras que terminan con dos o más signos de admiración seguidos, por ejemplo "!!!".

Debes imprimir tanto la palabra como la totalidad de signos de admiración que le siguen.

Indica cuántos resultados obtuviste: **26**



```python
matches = []

for doc in clean_docs:
    matches += re.findall(r'\b\w+!{2,}', doc)

print(matches)
print(len(matches))
```

*   **Pregunta 3.**  

Busca e imprime todas las palabras o cadena de 2 o más catacteres que están escritas totalmente en mayúsculas. Cada coincidencia debe ser una sola palabra (cadena de caracteres).

Indica cuántas palabras encontraste: **96**



```python
matches = []

for doc in clean_docs:
    matches += re.findall(r'\b[A-Z]{2,}\b', doc)

print(matches)
print(len(matches))
```

*   **Pregunta 4.**  

Busca e imprime los comentarios en donde todos los caracteres alfabéticos (letras) están en mayúsculas.

Cada coincidencia encontrada debe ser todo el comentario/enunciado.

Indica cuántos resultados obtuviste: **5**


```python
matches = []

for doc in clean_docs:
    if re.fullmatch(r'[^a-z]*[A-Z][^a-z]*', doc):
        matches.append(doc)

print(matches)
print(len(matches))
```

*   **Pregunta 5.**  

Busca e imprime todas las palabras que tengan una vocal acentuada, del tipo á, é, í, ó, ú.

Indica cuántos resultados obtuviste: **3**

```python
matches = []

for doc in clean_docs:
    matches += re.findall(r'\b\w*[áéíóú]\w*\b', doc)

print(matches)
print(len(matches))
```

*   **Pregunta 6.**  

Busca e imprime todas las cantidades numéricas monetarias, enteras o con decimales, que inician con el símbolo $\$$.

Indica cuántos resultados obtuviste: **8**

```python
matches = []

for doc in clean_docs:
    matches += re.findall(r'\$\d+(?:\.\d+)?', doc)

print(matches)
print(len(matches))
```

*   **Pregunta 7.**  

Busca e imprime todas las palabras que sean variantes de la palabra "love" (amor). No deberá importar si incluyen mayúsculas o minúsculas, o la manera en que esté conjugada o alguna otra variación que se haga con dicha palabra, como repetir letras por ejemplo.

Observa que deben ser variantes de la palabra love (amor) y no variantes de dicha cadena de caracteres.

Indica cuántos resultados obtuviste: **33**

```python
matches = []

for doc in clean_docs:
    matches += re.findall(r'\b[lL]+o+v+e+\w*\b', doc)

print(matches)
print(len(matches))
```

*   **Pregunta 8.**  

Busca e imprime todas las palabras que sean variantes de las palabras en inglés "so" y "good", y que tengan dos o más "o" en "so" y 3 o más "o" en good.

Observa que solo debes incluir variantes de estas palabras en inglés y no de dichas cadenas de caracteres.

Indica además cuántas encontraste: **4**


```python
matches = []

for doc in clean_docs:
    matches += re.findall(r'\bso{2,}\b|\bg o{3,}d\b'.replace(" ", ""), doc, re.IGNORECASE)

print(matches)
print(len(matches))
```

*   **Pregunta 9.**  

Busca e imprime todas las palabras que tengan una longitud mayor estrictamente a 12 caracteres alfabéticos.

No se consideran los signos de puntuación o caracteres especiales en la longitud de estas cadenas, solo caracteres alfabéticos en mayúsculas o minúsculas.

Indica la cantidad de palabras encontradas: **34**


```python
matches = []

for doc in clean_docs:
    matches += re.findall(r'\b[a-zA-Z]{13,}\b', doc)

print(matches)
print(len(matches))
```

*   **Pregunta 10.**  

Busca e imprime todas las palabras (caracteres alfabéticos) que inician con una letra mayúscula y terminan con una minúscula, pero donde además dicha palabra no sea la primera palabra del comentario (string).

Indica la cantidad de resultados obtenidos: **275**

```python
matches = []

for doc in clean_docs:
    matches += re.findall(r'(?<!^)\b[A-Z][a-z]+\b', doc)

print(matches)
print(len(matches))
```

*   **Pregunta 11.**  

Busca e imprime la secuencia de dos o más palabras que están separadas por un guion, "-", sin que tengan espacios en blanco entre ellas.

Por ejemplo "Go-Kart" sería válido, pero "Go  -Kart" o "Go  -  Kart" no lo serían, ya que incluyen algún espacio en blanco.

Indica la cantidad de resultados obtenidos: **19**

```python
matches = []

for doc in clean_docs:
    matches += re.findall(r'\b\w+(?:-\w+)+\b', doc)

print(matches)
print(len(matches))
```

*   **Pregunta 12.**  

Busca e imprime todas las palabras que terminan con la conjugación en inglés "ing" o "ed".

No deben importar mayúsculas o minúsculas.

Indica la cantidad de palabras que encontraste de cada una: **ing: 280 - ed: 339**

```python
ing = []
ed = []

for doc in clean_docs:
    ing += re.findall(r'\b\w+ing\b', doc, re.IGNORECASE)
    ed += re.findall(r'\b\w+ed\b', doc, re.IGNORECASE)

print("ing:", len(ing))
print("ed:", len(ed))
```

#**Parte 3. Proceso de limpieza.**

*   **Pregunta 13.**  

Ahora realiza un proceso de limpieza del corpus que incluya los siguientes procesos:

*   Solo se deben considerar caracteres alfabéticos, incluyendo algún tipo de acentuación. Es decir, se eliminan todos los signos de puntuación y caracteres especiales.
*   Todos los caracteres alfabéticos se transforman a minúsculas.
*   Se deben eliminar todos los espacios en blanco adicionales que se puedan encontrar en cada comentario.

Al finalizar dicho proceso de limpieza, imprime el resultado de los primeros 10 comentarios resultantes.
   




```python
clean_docs2 = []

for doc in clean_docs:
    # solo letras (incluye acentos)
    text = re.sub(r'[^a-zA-ZáéíóúÁÉÍÓÚ\s]', '', doc)

    # minúsculas
    text = text.lower()

    # quitar espacios extra
    text = re.sub(r'\s+', ' ', text).strip()

    clean_docs2.append(text)

print(clean_docs2[:10])
```

*   **Pregunta 14.**  

Supongamos que en un primer análisis básico consideramos como nuestro nuevo corpus el texto obtenido después del proceso de limpieza del ejercicio anterior.  

A partir de este nuevo corpus consideraremos cada palabra (cadena de caracteres) resultante como los "tokens" de nuestro estudio.

Y a partir de este documento tokenizado generamos nuestro diccionario de tokens.

¿Cuántos tokens diferentes tiene nuestro diccionario? **2051**


```python
tokens = []

for doc in clean_docs2:
    tokens += doc.split()

vocabulario = set(tokens)

print(len(vocabulario))
```

*   **Pregunta 15.**  

Finalmente, en este ejercicio definiremos nuestro conjunto de palabras denominado "stopwords", es decir, las palabras (cadenas de caracteres) que tienen frecuencias muy altas, pero que no aportan mucho valor para el significado/entendimiento de cada comentario.

Recuerda que ejemplos de stopwords son artículos, adverbios, conectivos, etcétera.

* **15a.** Con base a la lista de stopwords que se te proporciona, realiza un proceso de limpieza eliminando todas estas palabras del corpus obtenido en el ejercicio anterior. Imprime los primeros 5 comentarios resultantes.

```python
# Debes considerar la siguiente lista como tu conjunto de stopwords:
mis_stopwords = ['i', 'me', 'my', 'myself', 'we', 'our', 'ours', 'ourselves', 'you', 'your', 'yours', 'he', 'him', 'his', 'himself', 'she', 'her', 'hers', 'herself', 'it', 'its', 'itself', 'they', 'them', 'their', 'theirs', 'themselves', 'what', 'which', 'who', 'whom', 'this', 'that', 'these', 'those', 'am', 'is', 'are', 'was', 'were', 'be', 'been', 'being', 'have', 'has', 'had', 'having', 'do', 'does', 'did', 'doing', 'a', 'an', 'the', 'and', 'but', 'if', 'or', 'because', 'as', 'until', 'while', 'of', 'at', 'by', 'for', 'with', 'about', 'against', 'between', 'into', 'through', 'during', 'before', 'after', 'above', 'below', 'to', 'from', 'up', 'down', 'in', 'out', 'on', 'over', 'under', 'again', 'further', 'then', 'once', 'here', 'there', 'when', 'where', 'why', 'how', 'all', 'any', 'both', 'each', 'few', 'more', 'most', 'other', 'some', 'such', 'only', 'own', 'same', 'so', 'than', 'too', 'very', 's', 't', 'can', 'will', 'just', 'should', 'now', 'll']
```

```python
clean_docs3 = []

for doc in clean_docs2:
    palabras = doc.split()
    filtradas = [w for w in palabras if w not in mis_stopwords]
    clean_docs3.append(' '.join(filtradas))

print(clean_docs3[:5])
```

```python
tokens_final = []

for doc in clean_docs3:
    tokens_final += doc.split()

print("Tokens final:", len(tokens_final))

vocab_final = set(tokens_final)

print("Tokens únicos:", len(vocab_final))
```
