# 3. Creación de la base de conocimiento

Para simular una base documental real, los 20 currículums sintéticos se almacenarán en dos formatos:

- 16 documentos PDF
- 4 imágenes PNG

```python
!pip -q install fpdf2
!pip -q install pillow
```

```python
import os

os.makedirs("CVs", exist_ok=True)
os.makedirs("CVs/PDF", exist_ok=True)
os.makedirs("CVs/PNG", exist_ok=True)
```

```python
from fpdf import FPDF

def guardar_pdf(texto, archivo):

    pdf = FPDF()
    pdf.set_auto_page_break(auto=True, margin=15)

    pdf.add_page()
    pdf.set_font("Helvetica", size=11)

    # Limpiar caracteres incompatibles
    texto = texto.encode("latin-1", "replace").decode("latin-1")

    for linea in texto.split("\n"):

        linea = linea.strip()

        if linea == "":
            pdf.ln(4)
        else:
            pdf.multi_cell(
                w=180,
                h=7,
                text=linea
            )

    pdf.output(archivo)
```

```python
from PIL import Image, ImageDraw, ImageFont

def guardar_png(texto, archivo):

    ancho = 900
    alto = 1400

    imagen = Image.new("RGB", (ancho, alto), "white")

    draw = ImageDraw.Draw(imagen)

    fuente = ImageFont.load_default()

    draw.multiline_text(
        (30,30),
        texto,
        fill="black",
        font=fuente,
        spacing=5
    )

    imagen.save(archivo)
```

```python
# 1 Junior
# 2 Semi Senior
# 1 Senior

png_indices = [2,7,13,18]
```

```python
for i,(nivel,cv) in enumerate(cvs):

    nombre = f"CV_{i+1}"

    if i in png_indices:

        guardar_png(
            cv,
            f"CVs/PNG/{nombre}.png"
        )

    else:

        guardar_pdf(
            cv,
            f"CVs/PDF/{nombre}.pdf"
        )

print("Archivos generados correctamente.")
```

# 4. Preparación de los documentos

Los currículums fueron almacenados en PDF y PNG, pero el texto original se conservará en la memoria para construir la base vectorial.

```python
documentos = []

for i,(nivel,cv) in enumerate(cvs):

    if i in png_indices:
        formato = "PNG"
        archivo = f"CV_{i+1}.png"
    else:
        formato = "PDF"
        archivo = f"CV_{i+1}.pdf"

    documentos.append({
        "archivo": archivo,
        "formato": formato,
        "nivel": nivel,
        "texto": cv
    })
```

# 5. Construcción de la base vectorial

Para realizar la búsqueda inteligente de candidatos, cada currículum será transformado en un vector numérico (embedding).


```python
!pip -q install sentence-transformers
!pip -q install faiss-cpu
```

```python
from sentence_transformers import SentenceTransformer
import faiss
import numpy as np
```

```python
modelo = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
textos = [doc["texto"] for doc in documentos]

embeddings = modelo.encode(
    textos,
    convert_to_numpy=True
)

embeddings.shape
```

```python
dimension = embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)

index.add(embeddings)

print("Cantidad de CV indexados:", index.ntotal)
```

# 6. Búsqueda semántica

La descripción de la vacante también será convertida en un embedding.

```python
vacante = """
Machine Learning Engineer

Se busca un profesional con experiencia en:

Python

SQL

Machine Learning

TensorFlow

PyTorch

Azure

AWS

Git

Modelos predictivos

Análisis de datos

Trabajo en equipo

Resolución de problemas
"""
```

```python
embedding_vacante = modelo.encode(
    [vacante],
    convert_to_numpy=True
)
```

```python
k = 5

distancias, indices = index.search(
    embedding_vacante,
    k
)
```

```python
indices
```

```python
print("Top 5 candidatos\n")

for posicion, indice in enumerate(indices[0], start=1):

    doc = documentos[indice]

    print("="*70)
    print(f"Candidato {posicion}")
    print(f"Archivo: {doc['archivo']}")
    print(f"Nivel: {doc['nivel']}")
    print(f"Formato: {doc['formato']}")
```

```python
print("Resumen del candidato mejor evaluado:\n")

print(documentos[indices[0][0]]["texto"][:700])
```

```python
def obtener_habilidades(texto):

    conocidas = [
        "Python",
        "SQL",
        "TensorFlow",
        "PyTorch",
        "Azure",
        "AWS",
        "Git",
        "Docker",
        "Spark",
        "Power BI",
        "Pandas",
        "Scikit-Learn"
    ]

    encontradas = []

    for habilidad in conocidas:

        if habilidad.lower() in texto.lower():
            encontradas.append(habilidad)

    return encontradas
```

```python
for posicion, indice in enumerate(indices[0], start=1):

    doc = documentos[indice]

    print("="*70)

    print(f"Candidato {posicion}")

    print(f"Archivo: {doc['archivo']}")

    print(f"Nivel: {doc['nivel']}")

    habilidades = obtener_habilidades(doc["texto"])

    print("Habilidades encontradas:")

    print(", ".join(habilidades))
```

```python
ranking = []

for posicion, indice in enumerate(indices[0], start=1):

    doc = documentos[indice]

    ranking.append({

        "Posición": posicion,

        "Archivo": doc["archivo"],

        "Nivel": doc["nivel"],

        "Formato": doc["formato"],

        "Habilidades": ", ".join(
            obtener_habilidades(doc["texto"])
        )

    })

ranking_df = pd.DataFrame(ranking)

ranking_df
```
